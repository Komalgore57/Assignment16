## Task 1: Support Vector Machine (SVM)
1. Select a classification dataset.
2. Split into train and test sets.
3. Train an SVM classifier using SVC.
4. Experiment with:
- Linear kernel
- RBF kernel
5. Compare accuracy of both kernels.

[Download dataset ](https://www.kaggle.com/datasets/elakiricoder/gender-classification-dataset)

```python
# read dataset


# 1. Select a classification dataset.


path = "gender_classification_v7.csv"
df = pd.read_csv(path)
df.head()

lb =LabelEncoder()

Y = lb.fit_transform(df['gender'])

# 2. Split into train and test sets.

X = df.drop(columns= ['gender'])


# 3. Train an SVM classifier using SVC.

x_train, x_test , y_train , y_test = train_test_split(X,Y,test_size=0.2)

# 4. Experiment with:



# - Linear kernel

svc_linear = SVC(kernel='linear')
svc_linear.fit(x_train,y_train)
y_predict_linear = svc_linear.predict(x_test)


# - RBF kernel
svc_rbf = SVC(kernel='rbf')
svc_rbf.fit(x_train,y_train)
y_predict_rbf = svc_rbf.predict(x_test)

# 5. Compare accuracy of both kernels.


print("Accuracy Score ")
print(f"linear : {accuracy_score(y_true= y_test,y_pred=y_predict_linear)}")

print(f"rbc : {accuracy_score(y_true= y_test,y_pred=y_predict_rbf)}")
```

## Task 2: Decision Tree Algorithm

1. Train a Decision Tree Classifier.
2. Visualize the tree structure (depth limited).
3. Train models with:
- Low max_depth (underfitting)
- High max_depth (overfitting)
4. Compare train vs test accuracy.

```python
# 1. Train a Decision Tree Classifier.

tree = DecisionTreeClassifier(max_depth=7)
tree
# 2. Visualize the tree structure (depth limited).

## ASK 

# 3. Train models with:

# - Low max_depth (underfitting)
tree_low_depth = DecisionTreeClassifier(max_depth=1)
tree_low_depth.fit(x_train,y_train)
y_predict_low_depth = tree_low_depth.predict(x_test)


# - High max_depth (overfitting)
tree_high_depth = DecisionTreeClassifier(max_depth=7)
tree_high_depth.fit(x_train,y_train)
y_predict_high_depth = tree_high_depth.predict(x_test)
# 4. Compare train vs test accuracy.

print("Accuracy Score ")
print(f"Low Depth : {accuracy_score(y_true= y_test,y_pred=y_predict_low_depth)}")

print(f"High Depth : {accuracy_score(y_true= y_test,y_pred=y_predict_high_depth)}")
```
